# Clase 19. Swaps Collateralized

### 1. Introduccion

En entornos de mercado donde los contratos de derivados se colateralizan para eliminar el riesgo de crédito, la tasa de descuento deja de basarse en la curva de swaps y debe reflejar la remuneración del colateral pactado en el Credit Support Annex (CSA). Dado que ese colateral suele depositarse en efectivo, la tasa overnight (OIS) **que mejor captura el riesgo de un día** es la apropiada para descontar los flujos. Cuando, a partir de 2007, la diferencia entre la curva de swaps y la curva overnight se disparó, quedó patente la necesidad de adoptar un enfoque de *doble curve stripping*: una para proyectar los Tasas forwards y otra, la OIS, para descontar.

Un ejercicio empírico en escenario de tensión de crédito mostró que pasar a descontar con OIS altera significativamente la eficacia de las coberturas. En particular, los esquemas de “bucket hedging” (cubrir por tramos de vencimiento) superan a la cobertura mediante un único swap, y técnicas más sofisticadas prometen mejoras adicionales, aunque dependen de la liquidez de los mercados OIS. En conclusión, todo derivado colateralizado debe descontarse al Tasa de interés del colateral según el CSA, y las estrategias de cobertura han de contemplar de forma explícita la coexistencia de múltiples curvas de acceso al colateral o garantía.


### 2. Cash collateral 

Asumiendo que el Mark to Market value de un Swap es positivo para el banco y negativo para la contraparte (Fondo de Pensiones).

Si el swap está colateralizado en una sola divisa (EUR), los flujos se representan:

- 1) **The pension fund has to post collateral to the bank** and does so with EUR cash. 

- 2) **The bank pays EONIA** (For British pounds, SONIA would be the EONIA equivalent. In the US this would be the Federal funds rate/ OIS.) **as a fee over the collateral to the pension fund**. This is common practice in the swap market (Johannes and Sundaresan, 2007)

- 3) The bank finances this by **posting the collateral in a money market account*** on which **it receives EONIA**.

- 4) The collateral account is balanced daily and the reverse position is also possible. In this case, the MtM of the swap is positive to the pension fund and thus negative to the bank. The process is then reversed: 
    - The bank borrows money in the money market to post as collateral to the pension fund, for which it receives EONIA.
    - The latter is used to pay the fee to the money market counterparty.

*The money market is exemplary to illustrate **the revenue from the collateral for the bank**. 

In practice, **the bank could loan out the collateral cash to another bank for one day, for which it would receive EONIA over the amount**.

If the bank does not lend out the money because it needs it to **finance other activities, it has less need to borrow**.

*As the bank would have to pay EONIA over this loan, it ‘saves’ the interest burden due to the collateral posting.*
Regardless of what it does with the money, *the bank earns EONIA over the amount posted by the pension fund*.


### 3. Repaso y Justificación

El enfoque clásico valora un IRS como una cartera de contratos forward sobre el Tasa flotante. Bajo supuestos concretos de incumplimiento y riesgo de crédito de las contrapartes, Duffie y Singleton (1997) demuestran que la tasa swap equivale a la tasa de cupón de un bono “par” emitido bajo calidad LIBOR (AA) a lo largo de toda su vida (Hull, 2010a). A medida que el mercado se amplió y entraron contrapartes de distinta solvencia, surgieron técnicas para mitigar el riesgo de crédito en swaps (Johannes et al., 2007).

En caso de default, lo relevante para la contraparte es por un lado el interés acumulado neto hasta la fecha de incumplimiento y, por otro, el coste de reemplazo del swap, es decir, cuánto cuesta contratar uno nuevo con características equivalentes. Para reducir este riesgo de contraparte se recurre a la colateralización junto con el marcado a mercado diario (MtM), que minimiza la exposición neta y asegura que el valor del colateral cubra el coste de reposición al cierre de cada jornada.


**Contexto y definiciones**

El **swap spread** mide la diferencia entre la tasa fija de un interés rate swap (IRS) y el rendimiento de un bono soberano de igual plazo. Tras la crisis financiera, la compresión generalizada de estos spreads refleja, en gran parte, la adopción de colateralización y el uso de curvas OIS para descontar flujos en contratos garantizados. Sin embargo, dicha compresión no ha implicado una reducción homogénea de la volatilidad: ésta varía significativamente según la región y el vencimiento de los instrumentos .

**Dinámicas regionales de los swap spreads**

* **Estados Unidos (USD)**

  * *Compresión y territorio negativo*. Antes de 2008 los swap spreads en USD oscilaban en terreno positivo, en parte porque se descontaban con la curva LIBOR sin colateral. Tras la crisis, la compresión resultante de aplicar OIS discounting llevó los spreads casi a cero durante años. Desde la década actual, sin embargo, los swap spreads a plazos intermedios (p. ej., 5 y 10 años) se mantienen predominantemente **negativos** (desde 2015 en adelante), fenómeno asociado a cambios en la percepción de riesgo de los Treasuries y a incentivos regulatorios de balance sheet .
  * *Largo plazo consistente*. El swap spread a 30 años en USD lleva **negativo** desde fines de 2008, manteniendo una prima invertida que persiste ante la especial “convenience yield” de los Treasuries de corta duración y las presiones regulatorias sobre los bancos para encaje de largo plazo .

* **Eurozona (EUR)**

  * *Volatilidad acentuada*. A diferencia de EE. UU., los swap spreads en euros han experimentado al menos **tres picos marcados**:

    1. **Crisis soberana europea** (2010–2012).
    2. **Pandemia COVID-19** (marzo 2020).
    3. **Recuperación post-pandemia** (2021–2022), donde el spread subió de \~25 pb a más de 80 pb entre inicios de 2021 y finales de 2022.
  * *Tendencia a la paridad y territorio negativo*. En 2023–2024 el spread se ajustó bruscamente, estrechándose en 50 pb solo durante 2024 y alcanzando **cero** en noviembre de 2024. A partir de entonces, entró por primera vez en **territorio negativo**.

* **Japón (JPY)**

  * *Menor intensidad*. Aunque los swap spreads en JPY también muestran episodios de volatilidad, la magnitud de los movimientos ha sido significativamente inferior a la de EUR y USD, reflejando un mercado de Tasas más estable y de menor exposición a crisis de crédito soberano.

**Factores explicativos**

1. **Colateralización y OIS discounting**: el pasaje de LIBOR a OIS para descontar swaps colateralizados comprimió los spreads, pero introdujo sensibilidad a la curva overnight y creó nueva exposición al basis risk entre curvas. Muchas veces se utilizan bonos o treasurys como referencia de costo de oportunidad del colateral en la práctica. En Chile puede deberse en el caso de los fondos de pensiones a su trato con entidades internacionales (colateral en USD), por ejemplo CCS UF, SPC Peso, etc. cuyo colateral se transa en dólares.
2. **Riesgo soberano y “convenience yield”**: la percepción de riesgo de deuda pública (hasta incluir un ajuste de CDS en el modelo de valoración) puede justificar spreads negativos, al valorar los swaps como más seguros que ciertas emisiones de largo plazo .
3. **Regulación y balance sheets**: requerimientos de capital más exigentes para tenencia de bonos soberanos a largo plazo han incentivado a los dealers a tender a vender Treasuries e incrementar posiciones en swaps, presionando los spreads a terreno negativo.

### 4. Bond collateral

Contrary to banks, pension funds do not usually have a lot of cash at hand. They do typically have a large
asset allocation to government bonds, which can also be used as collateral. (Verheijen, 2009) Therefore,
pension funds specify their CSAs such that collateral on both sides is posted in the form of government
bonds.

En este punto es fundamental abordar el **costo de oportunidad del dinero**. Cuando se administra un fondo, usualmente el porcentaje de liquidez se mantiene a través de activos como bonos del tesoro, instrumentos de intermediación financiera IIF, treasury bills, etc. y se mantiene un nivel bajo (0,05%) en cuenta corriente. Por ende, ante los requerimientos de liquidez, muchas veces para entregar colateral y no quedar sin activos disponibles, o por la intención de mantener en cartera estos activos de corta duration, se debe conseguir este colateral en el mercado de repo's, y el dinero se obtiene desde el mercado money market. 

Por esta razón es que debemos ajustar el valor de nuestro contrato por el costo de oportunidad del colateral: Esto se conoce como FUNDING VALUE ADJUSTMENT (FVA).

En el caso que la contraparte quiera realizar el **"unwind"** del swap, entonces se devuelve el colateral, este colateral se devuelve al mercado repo, por lo cual nos devuelven la inversión a una tasa variable, y con esa tasa variable podemos devolver los fondos al mercado money market, consolidando las pérdidas o ganancias del contrato.

Esto genera un problema a diferencia del colateral en cash. Debido a que la Repo Market depende de cada Tasa de deudor (a quien se le presta dinero y entrega colateral en bonos). Por ende, el contrato CSA, debe especificar claramente el Tasa de bonos utilizado para definir la curva de descuento utilizada para el fondeo.

The most important result is that the funding cost of the collateral determines the discount curve used
to value the derivative.

### 5. Pricing & Valuation IRS

#### **5.1. Swaps de Tasas de interés sin colateral**

El enfoque tradicional para la valoración de un swap de Tasas de interés considera el swap como una cartera de contratos a plazo (forwards) sobre el Tasa de interés flotante subyacente. Se valora separadamente cada una de las piernas (legs) del swap. La pierna fija equivale a un bono de Tasa fijo sin intercambio del nocional. La pierna flotante equivale a un bono de Tasa variable sin intercambio de nocional. En la iniciación, el valor de ambas piernas es igual, lo que convierte al swap de Tasas de interés en un producto que no requiere inversión inicial. El Tasa fijo que anula el valor del contrato **Par Swap Rate** se cotiza en el mercado.

***Acá recuerden que lo que calculamos con nuestros factores descuentos proyectando los flujos futuros puede ser distinto a lo que se cotiza en el mercado, eventualmente si hay diferencias respecto a las curvas de descuento se pueden generar condiciones de arbitraje.*

La pierna flotante se valora descontando todos los flujos de caja. Dado que el pago futuro es desconocido, se utiliza el Tasa a plazo (forward rate) para su valoración. Si esto no fuera exacto, surgirían oportunidades de arbitraje.

Valoremos un swap simple a un año con nocional igual a 1, donde la pierna flotante paga el Tasa Libor a 6 meses cada seis meses y la pierna fija paga el Tasa fijo acordado anualmente. Esta es la práctica de mercado común en la zona euro.⁶

En primer lugar, algunas definiciones:

* $N$ = *Nocional del swap*
* $\alpha_{t,k}$ = *Fracción de año (daycount fraction) desde el tiempo $t$ hasta $k$*
* *La convención de cómputo de días (DCC) de la pierna fija es $30/360$, mientras que la de la pierna flotante es $ACT/360$.*
* *Alpha es la fracción de año entre dos fechas, calculada según la DCC correspondiente.*
* $z_k$ = *Tasa cero Euribor con vencimiento $k$*
* $D_k$ = *Factor de descuento Euribor con vencimiento $k$, dado por*

  $$
    D_k = \frac{1}{1 + z_k \,\alpha_{t,k}}  
  \tag{1}  
  $$
* $F_{t,k}$ = *Tasa Forward Euribor desde $t$ hasta $k$, dado por*

  $$
    F_{t,k} = \frac{1}{\alpha_{t,k}}\biggl(\frac{D_t}{D_k} - 1\biggr)  
  \tag{2}  
  $$
* $K_n$ = *Tasa fijo del swap Euribor con vencimiento $n$, observado en el mercado.*

*Cuando se utilice alguno de los símbolos anteriores para indicar un Tasa de swap sobre índice overnight (relacionado con EONIA), se añadirá el superíndice OIS.*





Ahora valoraremos un swap a un año, que tiene dos piernas, cada una con una secuencia de flujos de caja que hay que descontar. La pierna fija paga anualmente y, por tanto, un único flujo fijo, cuyo valor presente es:

$$
PV(\text{fija}) = D_{1\text{y}} \,\alpha_{0,1\text{y}} \,K_{1\text{y}}
$$

Para valorar la pierna flotante, primero definimos los Tasas a plazo en función de los factores de descuento. En un mundo de capitalización simple, habría arbitraje si no se cumpliera, para el caso a dos años, la siguiente relación:

$$
(1 + z_1\,\alpha_{0,1})\;\bigl(1 + F_{1,2}\,\alpha_{1,2}\bigr)
\;=\;
1 + z_2\,\alpha_{0,2}
$$

El Tasa a plazo de 1 a 2 años se obtiene fácilmente como:

$$
F_{1,2}
\;=\;
\biggl(\frac{1 + z_2\,\alpha_{0,2}}{1 + z_1\,\alpha_{0,1}} - 1\biggr)\,\frac{1}{\alpha_{1,2}}
$$

A continuación, invertimos (1) para expresar los Tasas cero en función de los factores de descuento:

$$
D_k = \frac{1}{1 + z_k\,\alpha_{t,k}}
\quad\Longrightarrow\quad
z_k = \biggl(\frac{1}{D_k} - 1\biggr)\,\frac{1}{\alpha_{0,k}}
$$

Podemos usar este resultado para reescribir $F_{1,2}$:

$$
F_{1,2}
=
\Biggl(\Bigl(1 + \bigl(\tfrac{1}{D_2} - 1\bigr)\tfrac{1}{\alpha_{0,2}}\Bigr)\alpha_{0,2}
\;\Bigm/\;
\Bigl(1 + \bigl(\tfrac{1}{D_1} - 1\bigr)\tfrac{1}{\alpha_{0,1}}\Bigr)\alpha_{0,1}
\;-\;1\Biggr)\;
\tfrac{1}{\alpha_{1,2}}
$$

$$
F_{1,2}
=
\frac{D_1}{D_2}\,\frac{1}{\alpha_{1,2}}
\;-\;
\biggl(\frac{D_1}{D_2} - 1\biggr)\frac{1}{\alpha_{1,2}}
$$

Este resultado se generaliza en (2). Es necesario para valorar la pierna flotante del swap a un año. La pierna flotante se valora descontando los dos flujos: el pago a 6 meses y el pago final al año:

$$
PV(\text{flt})
= \alpha_{0,6\text{m}}\,F_{0,6\text{m}}\,D_{6\text{m}}
+ \alpha_{6\text{m},1\text{y}}\,F_{6\text{m},1\text{y}}\,D_{1\text{y}}
$$

**Reescribiendo los Tasas a plazo usando (2) obtenemos:**

$$
PV(\text{flt})
= \alpha_{0,6\text{m}}\,
\frac{1}{\alpha_{0,6\text{m}}}\Bigl(\tfrac{D_0}{D_{6\text{m}}}-1\Bigr)\,D_{6\text{m}}
\;+\;
\alpha_{6\text{m},1\text{y}}\,
\frac{1}{\alpha_{6\text{m},1\text{y}}}\Bigl(\tfrac{D_{6\text{m}}}{D_{1\text{y}}}-1\Bigr)\,D_{1\text{y}}
$$

$$
PV(\text{flt})
= \Bigl(\tfrac{D_0}{D_{6\text{m}}}-1\Bigr)D_{6\text{m}}
+ \Bigl(\tfrac{D_{6\text{m}}}{D_{1\text{y}}}-1\Bigr)D_{1\text{y}}
$$

$$
PV(\text{flt})
= D_0 - D_{6\text{m}} + D_{6\text{m}} - D_{1\text{y}}
$$

$$
PV(\text{flt})
= D_0 - D_{1\text{y}}
$$

Este resultado también se mantiene para swaps plurianuales, lo que simplifica notablemente su valoración.

Para un vencimiento a dos años, hay dos pagos fijos; el primero se descuenta con el factor a un año, el cual depende del Tasa a un año según (1). Así, el valor de un swap a dos años está influido por dos Tasas, mientras que un instrumento cero cupón a dos años sólo está expuesto al Tasa a dos años.

Los resultados se generalizan como sigue⁷:

$$
PV(\text{fija})
= N \,K_n \sum_{k=1}^{n} D_k \,\alpha_{t,k}
\tag{4}
$$

$$
PV(\text{flotante})
= N \sum_{k=1}^{n} D_k \,\alpha_{t,k}\,F_{t,k}
\tag{5}
$$

**Que se reduce a:**

$$
PV(\text{flotante})
= N\,(D_0 - D_n)
\tag{6}
$$

*El valor de un swap receptor (receiver) es la pierna fija menos la pierna flotante, mientras que el de un swap pagador (payer) es el opuesto.*


* La pierna fija y la flotante se valoran por separado y no se suman para obtener el valor de un swap, ya que no queda claro si nos referimos a un swap pagador o receptor. Si en esta tesis se menciona “el valor de un swap”, se trata de un swap receptor (el más utilizado por los fondos de pensiones), y dicho valor es la pierna fija menos la pierna flotante.*


```python
# Verificación de resultados y cálculo de Mark-to-Market

def discount_factor(z: float, alpha: float) -> float:
    """Calcula el factor de descuento D = 1 / (1 + z * alpha)."""
    return 1.0 / (1.0 + z * alpha)

def forward_rate(D_t: float, D_k: float, alpha: float) -> float:
    """Calcula el Tasa a plazo F = (D_t/D_k - 1) / alpha."""
    return (D_t / D_k - 1.0) / alpha

def pv_fixed_leg(zero_rates: dict, daycount: dict, K: float) -> float:
    """PV de la pierna fija con un único pago anual."""
    alpha = daycount['0-1y']
    D_1y = discount_factor(zero_rates['1y'], alpha)
    return D_1y * alpha * K

def pv_floating_leg(zero_rates: dict, daycount: dict) -> float:
    """PV de la pierna flotante: 1 - D_1y."""
    D_1y = discount_factor(zero_rates['1y'], daycount['0-1y'])
    return 1.0 - D_1y

def par_swap_rate(zero_rates: dict, daycount: dict) -> float:
    """Tasa par K = (1 - D_1y) / (D_1y * alpha)."""
    alpha = daycount['0-1y']
    D_1y = discount_factor(zero_rates['1y'], alpha)
    return (1.0 - D_1y) / (D_1y * alpha)

def mark_to_market(zero_rates: dict, daycount: dict, K: float) -> float:
    """
    Mark-to-market para un swap receptor (receiver):
    MTM = PV_fija(K) - PV_flot
    """
    return pv_fixed_leg(zero_rates, daycount, K) - pv_floating_leg(zero_rates, daycount)


```

```python

# Parámetros de mercado
zero_rates = {'6m': 0.035, '1y': 0.038}
daycount   = {'0-6m': 0.5, '6m-1y': 0.5, '0-1y': 1.0}

K_example = 0.038

# Cálculos
D_6m = discount_factor(zero_rates['6m'], daycount['0-6m'])
D_1y = discount_factor(zero_rates['1y'], daycount['0-1y'])
pv_fija = pv_fixed_leg(zero_rates, daycount, K_example)
pv_flot = pv_floating_leg(zero_rates, daycount)
K_par = par_swap_rate(zero_rates, daycount)
mtm   = mark_to_market(zero_rates, daycount, K_example)


F_6m_1y = forward_rate(D_6m, D_1y, daycount['6m-1y'])


print(f"Factor descuento 6m: {D_6m:.6f}")
print(f"Factor descuento 1y: {D_1y:.6f}")
print(f"Forward rate 6m→1y: {F_6m_1y:.4%}")
print(f"PV pierna fija (K={K_example:.2%}): {pv_fija:.6f}")
print(f"PV pierna flotante: {pv_flot:.6f}")
print(f"Swap par (K_par): {K_par:.4%}")
```

```python


# Verificación de consistencia
assert abs(K_par - zero_rates['1y']) < 1e-8, "Error: el Tasa par no coincide con la tasa cero anual."

# Salida de resultados
print(f"Verificación exitosa: el Tasa par ({K_par:.4%}) coincide con la tasa cero anual ({zero_rates['1y']:.4%}).")
print(f"MTM (receiver) con K={K_example:.2%}: {K_par} {mtm:.6f}")
```

### **5.2. Overnight index swap (OIS)**

Un Overnight index swap (OIS) es un producto colateralizado que intercambia un tipo fijo por el tipo de interés overnight. La pierna fija se valora de forma muy similar a un swap sin colateralización sobre una curva única, salvo que la curva subyacente es ahora la curva OIS. La pierna flotante es distinta, ya que paga el tipo diario en lugar del tipo semestral. Sin embargo, el pago real sólo se efectúa anualmente. Esto implica que las fechas de reseteo no coinciden con las fechas de pago, y por tanto los intereses se capitalizan entre cada reseteo.

La ecuación (4) puede reescribirse utilizando factores de descuento OIS y un tipo fijo extraído de la curva OIS, lo que conduce a:

$$
PV(\text{fija}) = N \cdot K_n^{\mathrm{OIS}} \cdot \sum_{k=1}^{n} D_k^{\mathrm{OIS}} \cdot \alpha_{t,k}
\tag{7}
$$





La pierna flotante sigue un enfoque diferente. Las realizaciones diarias del tipo OIS flotante van acumulando el interés Eonia, lo que conduce a:

$$
PV(\text{flotante}) = D_t^{\mathrm{OIS}} \;\Bigl(\prod_{k=1}^{n}(1 + F_{t,k}\,\alpha_{t,k}) \;-\; 1\Bigr)
$$

Reescribiendo los tipos a plazo de forma análoga a (2), obtenemos:

$$
PV(\text{flotante}) = D_t^{\mathrm{OIS}} \;\Bigl(\prod_{k=1}^{n}\Bigl(1 + \tfrac{1}{\alpha_{t,k}}\bigl(\tfrac{D_t^{\mathrm{OIS}}}{D_k^{\mathrm{OIS}}}-1\bigr)\,\alpha_{t,k}\Bigr) - 1\Bigr)
$$

$$
PV(\text{flotante}) = D_t^{\mathrm{OIS}} \;\Bigl(\tfrac{D_0^{\mathrm{OIS}}}{D_n^{\mathrm{OIS}}} - 1\Bigr)
$$

Para un swap OIS a un año:

$$
PV(\text{flotante})
= D_0^{\mathrm{OIS}}\;\Bigl(\tfrac{D_0^{\mathrm{OIS}}}{D_1^{\mathrm{OIS}}}\cdot\tfrac{D_1^{\mathrm{OIS}}}{D_2^{\mathrm{OIS}}}\cdot\cdots\cdot\tfrac{D_{364}^{\mathrm{OIS}}}{D_{365}^{\mathrm{OIS}}} - 1\Bigr)
$$

Lo cual se reduce a:

$$
PV(\text{flotante}) = D_0^{\mathrm{OIS}}\;\Bigl(\tfrac{D_0^{\mathrm{OIS}}}{D_{365}^{\mathrm{OIS}}} - 1\Bigr)
\quad\longrightarrow\quad
PV(\text{flotante}) = D_0^{\mathrm{OIS}} - D_{365}^{\mathrm{OIS}}
$$



Que se puede generalizar como:

$$
PV(\text{flotante}) = N \;\bigl(D_0^{\mathrm{OIS}} - D_n^{\mathrm{OIS}}\bigr)
\tag{8}
$$

Esto es idéntico a (6). Por tanto, concluimos que la metodología de valoración de un swap OIS coincide con la de un swap Euribor sin colateralización, salvo que se emplea una curva distinta.


*⁸ Puede parecer que hay 365 reseteos en un año, pero en realidad sólo ocurren en días hábiles; en fines de semana y festivos se capitaliza al tipo diario del último día hábil.*



```python
from typing import Dict

def discount_factor(z: float, alpha: float) -> float:
    """
    Factor de descuento OIS: D = 1 / (1 + z * alpha)

    z     – tasa cero OIS para el periodo
    alpha – fracción de año (por ejemplo, 1/365 para un día, 1 para un año)
    """
    return 1.0 / (1.0 + z * alpha)

def pv_fixed_leg_ois(zero_rates: Dict[str, float],
                     daycount: Dict[str, float],
                     K: float) -> float:
    """
    Valor presente de la pierna fija OIS:
      PV_fixed = N * K * sum_{k=1..n} D_k^{OIS} * alpha_{t,k}
    En nuestro caso de 1 año y pago anual, sólo hay un término:
      PV_fixed =    D_{365}^{OIS} * alpha_{0,365} * K
    """
    alpha = daycount['0-1y']
    D_1y = discount_factor(zero_rates['1y'], alpha)
    return D_1y * alpha * K

def pv_floating_leg_ois(zero_rates: Dict[str, float],
                        daycount: Dict[str, float]) -> float:
    """
    Valor presente de la pierna flotante OIS:
      PV_flt = N * (D_0^{OIS} - D_n^{OIS})
    Con D_0 = 1:
      PV_flt = 1 - D_{365}^{OIS}
    """
    D_1y = discount_factor(zero_rates['1y'], daycount['0-1y'])
    return 1.0 - D_1y

def par_swap_rate_ois(zero_rates: Dict[str, float],
                      daycount: Dict[str, float]) -> float:
    """
    Tipo par OIS K^{OIS} que anula el swap:
      K = (D_0 - D_n) / sum D_k * alpha_{t,k}
    En pago anual único:
      K = (1 - D_{365}) / (D_{365} * alpha_{0,365})
    """
    alpha = daycount['0-1y']
    D_1y = discount_factor(zero_rates['1y'], alpha)
    return (1.0 - D_1y) / (D_1y * alpha)

def mark_to_market_ois(zero_rates: Dict[str, float],
                       daycount: Dict[str, float],
                       K: float) -> float:
    """
    Mark-to-market del swap (receiver):
      MTM = PV_fixed(K) - PV_flt
    """
    pv_fix = pv_fixed_leg_ois(zero_rates, daycount, K)
    pv_flt = pv_floating_leg_ois(zero_rates, daycount)
    return pv_fix - pv_flt

# Tasas cero OIS: diaria (no la usamos explícitamente) y anual
zero_rates = {
    '1y': 0.015  # por ejemplo, 1.50% anual OIS
}
# Fracciones de año según convención ACT/365
daycount = {
    '0-1y': 1.0
}
# Tipo fijo pactado en el swap (ejemplo 1.40%)
K_example = 0.014

# Cálculos
pv_fix = pv_fixed_leg_ois(zero_rates, daycount, K_example)
pv_flt = pv_floating_leg_ois(zero_rates, daycount)
K_par = par_swap_rate_ois(zero_rates, daycount)
mtm   = mark_to_market_ois(zero_rates, daycount, K_example)

print(f"PV pierna fija OIS  : {pv_fix:.6f}")
print(f"PV pierna flotante OIS: {pv_flt:.6f}")
print(f"Tipo par OIS         : {K_par:.4%}")
print(f"MTM (receiver, K={K_example:.2%}): {mtm:.6f}")
```

### **5.3. Swap de tasas de interés con colateral en efectivo (Cash Collateralized IRS)**

Cuando un *swap* normal (basado en Euribor) está colateralizado con efectivo, la tasa de descuento Euribor deja de ser aplicable, como se ha discutido extensamente anteriormente. La naturaleza libre de riesgo que surge del colateral se refleja en la tasa de descuento libre de riesgo, es decir, la tasa OIS. Por lo tanto, las fórmulas de valoración previamente presentadas \[(4) y (6)] ya no son válidas, dado que dichas expresiones sólo son correctas bajo un único esquema de curvas (*single-curve framework*).

Cuando se utiliza la curva OIS para descontar flujos en un *swap* Euribor colateralizado, es necesario combinar la valoración tradicional de los *swaps* Euribor con el carácter libre de riesgo implícito en los *OIS*. Para el *fixed leg*, esto implica combinar las fórmulas (4) y (7), lo que resulta en:

$$
PV(\text{fixed}) = N \cdot K_n \cdot \sum_{k=1}^{n} D_k^{OIS} \cdot \alpha_{t,k} \tag{9}
$$

Los flujos fijos determinados a partir de la curva Euribor son descontados utilizando la curva OIS, ya que el colateral transforma el *swap* en un instrumento libre de riesgo. La misma lógica se aplica al *floating leg*, combinando la fórmula (5) con el descuento bajo OIS (en este caso, derivado de la EONIA):

$$
PV(\text{flt}) = N \cdot \sum_{k=1}^{n} D_k^{OIS} \cdot \alpha_{t,k} \cdot F_{t,k} \tag{10}
$$

Los *forward rates* utilizados para valorar el *floating leg* continúan siendo tasas Euribor, mientras que el descuento de estos flujos se realiza con la curva OIS. Es importante notar que la simplificación obtenida en (6) ya no es válida, dado que los factores de descuento utilizados para calcular las tasas *forward* provienen de la curva Euribor, mientras que los flujos son descontados con OIS. Esto introduce una descomposición entre la curva de proyección (*forward curve*) y la curva de descuento (*discount curve*), elemento central en el enfoque de valoración con curvas múltiples (*multi-curve framework*).



```python
# Modelo Python para Cash-Collateralized IRS con curvas múltiples

from typing import Dict

# 1. Definición de curvas cero y fracciones de año
euribor_zero: Dict[str, float] = {
    '6m': 0.035,  # 3.50% semestral proyectado
    '1y': 0.038   # 3.80% anual proyectado
}
ois_zero: Dict[str, float] = {
    '6m': 0.015,  # 1.50% semestral de descuento
    '1y': 0.017   # 1.70% anual de descuento
}

daycount_fixed: Dict[str, float] = {
    '0-1y': 1.0   # un pago fijo anual
}
daycount_float: Dict[str, float] = {
    '0-6m': 0.5,  # semestral flotante
    '6m-1y': 0.5
}

# 2. Funciones puras
def discount_factor(z: float, alpha: float) -> float:
    """Factor de descuento: D = 1 / (1 + z * alpha)."""
    return 1.0 / (1.0 + z * alpha)

def forward_rate(D_t: float, D_k: float, alpha: float) -> float:
    """Tipo a plazo: F = (D_t/D_k - 1) / alpha."""
    return (D_t / D_k - 1.0) / alpha

# 3. Cálculo del swap colateralizado
def example_swap_collateralized():
    N = 1.0           # nocional
    K_example = 0.037 # 3.70% tipo fijo pactado
    
    # 3.1 Factores de descuento OIS
    D_ois_6m = discount_factor(ois_zero['6m'], daycount_float['0-6m'])
    D_ois_1y = discount_factor(ois_zero['1y'], daycount_fixed['0-1y'])
    
    # 3.2 Factores de descuento proyectados (Euribor)
    D_proj_6m = discount_factor(euribor_zero['6m'], daycount_float['0-6m'])
    D_proj_1y = discount_factor(euribor_zero['1y'], daycount_fixed['0-1y'])
    
    # 3.3 Forward rates Euribor
    F_0_6m = forward_rate(1.0, D_proj_6m, daycount_float['0-6m'])
    F_6m_1y = forward_rate(D_proj_6m, D_proj_1y, daycount_float['6m-1y'])
    
    # 3.4 Valor presente piernas
    PV_fixed = N * K_example * D_ois_1y * daycount_fixed['0-1y']
    PV_float = N * (F_0_6m * daycount_float['0-6m'] * D_ois_6m +
                    F_6m_1y * daycount_float['6m-1y'] * D_ois_1y)
    
    # 3.5 Tipo par colateralizado y MTM (receiver)
    par_rate = PV_float / (D_ois_1y * daycount_fixed['0-1y'])
    mtm = PV_fixed - PV_float
    
    # Resultados
    print("=== Cash-Collateralized IRS ===")
    print(f"Factor descuento OIS 6m: {D_ois_6m:.6f}")
    print(f"Factor descuento OIS 1y: {D_ois_1y:.6f}")
    print(f"Forward Euribor 0→6m: {F_0_6m:.4%}")
    print(f"Forward Euribor 6m→1y: {F_6m_1y:.4%}")
    print(f"PV pierna fija: {PV_fixed:.6f}")
    print(f"PV pierna flotante: {PV_float:.6f}")
    print(f"Swap par rate collateralizado: {par_rate:.4%}")
    print(f"Mark-to-Market (receiver): {mtm:.6f}")

# Ejecutar el ejemplo
example_swap_collateralized()
```

```python
import numpy as np
import pandas as pd
from datetime import date

def generate_schedule(maturity_years: float, freq: int) -> list:
    """Devuelve los tiempos (en años) de pago para una frecuencia dada."""
    accrual = 1.0 / freq
    n_periods = int(maturity_years * freq)
    return [round((i + 1) * accrual, 10) for i in range(n_periods)]

def get_zero_rate(curve_times: list, curve_rates: list, t: float) -> float:
    """Interpolación lineal de la curva cero para un tiempo t."""
    return float(np.interp(t, curve_times, curve_rates))

def simple_discount(z: float, t: float) -> float:
    """Factor de descuento simple: D(0,t) = 1 / (1 + z * t)."""
    return 1.0 / (1.0 + z * t)

def forward_rate_simple(D_t: float, D_k: float, alpha: float) -> float:
    """Tipo a plazo simple: F = (D_t/D_k - 1) / alpha."""
    return (D_t / D_k - 1.0) / alpha

def value_cash_coll_swap(
    maturity_years: float,
    N: float,
    K: float,
    fixed_freq: int,
    float_freq: int,
    proj_curve: dict,
    disc_curve: dict,
    valuation_date: date
):
    """Valora un Cash-Collateralized IRS en un framework de curvas múltiples."""
    if valuation_date is None:
        valuation_date = date.today()
    
    # Generar cronograma
    fixed_times = generate_schedule(maturity_years, fixed_freq)
    float_times = generate_schedule(maturity_years, float_freq)
    accrual_fixed = 1.0 / fixed_freq
    accrual_float = 1.0 / float_freq
    
    rows = []
    # Pierna fija
    for t_end in fixed_times:
        D_disc = simple_discount(get_zero_rate(disc_curve['times'], disc_curve['rates'], t_end), t_end)
        cash_flow = N * accrual_fixed * K
        pv = cash_flow * D_disc
        rows.append({
            'ValDate': valuation_date,
            'Leg': 'Fixed',
            't_start': round(t_end - accrual_fixed, 4),
            't_end': t_end,
            'Accrual': accrual_fixed,
            'Rate': K,
            'CashFlow': cash_flow,
            'DiscFactor': D_disc,
            'PV': pv
        })
    # Pierna flotante
    for i, t_end in enumerate(float_times):
        t_start = 0.0 if i == 0 else float_times[i-1]
        # Obtener D_proj en t_start y t_end
        D_proj_t0 = simple_discount(get_zero_rate(proj_curve['times'], proj_curve['rates'], t_start), t_start)
        D_proj_t1 = simple_discount(get_zero_rate(proj_curve['times'], proj_curve['rates'], t_end), t_end)
        rate = forward_rate_simple(D_proj_t0, D_proj_t1, accrual_float)
        cash_flow = N * accrual_float * rate
        D_disc = simple_discount(get_zero_rate(disc_curve['times'], disc_curve['rates'], t_end), t_end)
        pv = cash_flow * D_disc
        rows.append({
            'ValDate': valuation_date,
            'Leg': 'Floating',
            't_start': round(t_start, 4),
            't_end': t_end,
            'Accrual': accrual_float,
            'Rate': rate,
            'CashFlow': cash_flow,
            'DiscFactor': D_disc,
            'PV': pv
        })
    
    df = pd.DataFrame(rows)
    # Totales y métricas
    pv_fix = df[df['Leg']=='Fixed']['PV'].sum()
    pv_flt = df[df['Leg']=='Floating']['PV'].sum()
    annuity = (df[df['Leg']=='Fixed']['DiscFactor'] * df[df['Leg']=='Fixed']['Accrual']).sum()
    par_rate = pv_flt / annuity
    mtm = pv_fix - pv_flt
    
    totals = pd.DataFrame([{
        'ValDate': valuation_date,
        'PV_Fixed': pv_fix,
        'PV_Floating': pv_flt,
        'ParRate': par_rate,
        'MTM_Receiver': mtm
    }])
    
    # Mostrar tablas
    print("Desarrollo de flujos y VP por periodo", df)
    return totals, df

# Parámetros de ejemplo
proj_curve = {'times': [0.5, 1.0], 'rates': [0.035, 0.038]}
disc_curve = {'times': [0.5, 1.0], 'rates': [0.015, 0.017]}

# Llamada
totals, df = value_cash_coll_swap(
    maturity_years=1.0,
    N=1.0,
    K=0.037,
    fixed_freq=1,
    float_freq=2,
    proj_curve=proj_curve,
    disc_curve=disc_curve,
    valuation_date=date(2025, 7, 13)
)

```

## **6. Construcción de la curva**

En esta sección se explica la construcción de las curvas de cero cupón. Primero se abordan algunos aspectos relacionados.

### **6.1. Instrumentos de curva**

Como se aprecia en el apéndice, los dos instrumentos más cortos utilizados para construir la curva Euribor son los instrumentos al contado a seis meses y a un año, aunque existen instrumentos líquidos de, por ejemplo, una semana y tres meses. Sin embargo, dado que el extremo corto carece de importancia en esta tesis debido a los horizontes largos, solo se eligen estos dos instrumentos cupón cero para construir la curva.

Los instrumentos swap utilizados tienen todos un tenor flotante de seis meses. La razón es que muchos autores, entre ellos Bianchetti (2009), subrayan la importancia de usar instrumentos de curva homogéneos en el tenor de la tasa subyacente. Esto significa que el lado flotante del swap debe ser igual para todos los instrumentos. Características de mercado como liquidez y primas por riesgo de crédito son muy diferentes para distintos tenores de tasa flotante. El mercado de swaps fijo versus flotante a 3 meses tiene una dinámica distinta al mercado de swaps fijo versus flotante a 6 meses. Esto no siempre fue así, pero la crisis «ha segmentado el mercado en subáreas correspondientes a distintos tenores de tasa subyacente», según Bianchetti (2010).

Para la curva Eonia, el tenor flotante de todos los instrumentos es siempre de un solo día, de modo que no pueden surgir problemas de este tipo. Podrían añadirse más instrumentos al extremo corto, pero el efecto en un portafolio de alta duración sería despreciable. Para mantener la coherencia, se usan los mismos instrumentos que para la curva Euribor, salvo los tres instrumentos de mayor vencimiento, que no están disponibles para Eonia. Esto se discute en la siguiente subsección.





### **6.2. Construcción de la curva**

Con el conocimiento de las subsecciones anteriores, se pueden analizar los swaps de mercado. En particular, el hecho de que su valor sea cero al inicio ofrece oportunidades en combinación con una cotización de mercado. Para la valoración de un gran conjunto de flujos de caja con diferentes vencimientos, se requiere una curva de ceros que permita derivar el factor de descuento para cada flujo. El extremo corto de la curva, hasta un año, se construye a partir de instrumentos al contado y por tanto se conoce. Sin embargo, para vencimientos más largos, el mercado de swaps es más líquido, lo que lo hace una mejor elección. Para extraer la tasa cero de un swap, podemos reescribir las ecuaciones (4) y (5), dado que el valor de la pata flotante debe igualar al de la pata fija:

$$
N \cdot K_T \cdot \sum_{k=1}^{n} D_k \cdot \alpha_{t,k} - N \cdot \sum_{k=1}^{n} D_k \cdot \alpha_{t,k} \cdot F_{t,k} = 0
$$

$$
K_T \cdot \sum_{k=1}^{n} D_k \cdot \alpha_{t,k} - (D_0 - D_n) = 0
$$

Dado que el instrumento a un año es al contado, podemos derivar $D_{1y}$ usando (1). La única incógnita, entonces, es la cotización del swap a un año:

$$
K_{1y} \cdot D_{1y} \cdot \alpha_{0,1} - (1 - D_{1y}) = 0
$$

De esta ecuación podemos resolver $K_{1y}$ usando (1), ya que es la única variable desconocida. El tipo fijo, por supuesto, también podría extraerse de la cotización del swap a un año, pero este método es preferido por razones de liquidez. El valor de un swap a dos años es:

$$
K_{2y} \cdot \bigl(D_{1y} \cdot \alpha_{0,1} + D_{2y} \cdot \alpha_{1,2}\bigr) - (1 - D_{2y}) = 0
$$

Como ya hemos calculado $D_{1y}$ y $K_{2y}$ es la tasa swap cotizada en el mercado para el instrumento a dos años, solo queda un parámetro desconocido: $D_{2y}$. Podemos resolver $D_{2y}$:

$$
K_{2y} \cdot \bigl(D_{1y} \cdot \alpha_{0,1} + D_{2y} \cdot \alpha_{1,2}\bigr) = 1 - D_{2y}
$$

$$
K_{2y} \cdot D_{1y} \cdot \alpha_{0,1} + K_{2y} \cdot D_{2y} \cdot \alpha_{1,2} + D_{2y} = 1
$$

$$
D_{2y} \cdot (1 + K_{2y} \cdot \alpha_{1,2}) = 1 - K_{2y} \cdot D_{1y} \cdot \alpha_{0,1}
$$

$$
D_{2y} = \frac{1 - K_{2y} \cdot D_{1y} \cdot \alpha_{0,1}}{1 + K_{2y} \cdot \alpha_{1,2}}
$$

Esto se puede generalizar a:

$$
D_n = \frac{1 - K_n \cdot \left( \sum_{k=1}^{n-1} D_k \cdot \alpha_{t,k} \right)}{1 + K_n \cdot \alpha_{t,n}} \tag{13}
$$

De este modo, podemos resolver paso a paso todos los factores de descuento de los instrumentos de la curva. Usando (1), podemos luego convertir los factores de descuento en tasas cero para obtener la curva requerida. El proceso descrito se conoce generalmente como **bootstrapping**.



#### **Uso de la curva**

Una vez construida una curva, cualquier flujo de caja en el futuro puede descontarse adecuadamente. Para flujos con vencimiento superior a 50 años, se asume la tasa cero a 50 años, lo que implica una extrapolación plana de la tasa cero. Esto se hace por falta de alternativas mejores y por la observación de que las tasas a muy largo plazo se correlacionan fuertemente.

También se puede construir una curva Eonia usando el procedimiento anterior, ya que los swaps OIS dependen de una única curva aunque los contratos estén colateralizados. Las cotizaciones de swaps y forwards se obtienen efectivamente de la curva Eonia, que a su vez se usa para el descuento. Recuerde que los flujos de las patas se descuentan de manera similar, salvo por la curva utilizada. Inversamente, las dos curvas pueden construirse con el mismo método (adivinen... sí, *Bootstraping*), pero con diferentes instrumentos.



### **6.3. Ajuste forward**

Dado que las cotizaciones de swaps Euribor asumen colateralización, no se puede construir una curva Euribor realista de esta manera.

Esto se debe a los factores de descuento OIS en la fórmula de valoración. Sin embargo, estos factores OIS no son desconocidos; pueden derivarse de la curva Eonia construida primero. Dadas las cotizaciones de swaps Euribor, se tiene entonces:

$$
N \cdot K_n \cdot \sum_{k=1}^{n} D_k^{OIS} \cdot \alpha_{t,k} - N \cdot \sum_{k=1}^{n} D_k^{OIS} \cdot \alpha_{t,k} \cdot F_{t,k} = 0
$$

Para un swap a dos años esto se convierte en:

$$
K_{2y} \cdot \bigl(D_{1y}^{OIS} \cdot \alpha_{0,1} + D_{2y}^{OIS} \cdot \alpha_{1,2}\bigr) 
- \bigl(F_{0,6m} \cdot \alpha_{0,6m} \cdot D_{6m}^{OIS} + F_{6m,1y} \cdot \alpha_{6m,1y} \cdot D_{1y}^{OIS} + F_{1y,18m} \cdot \alpha_{1y,18m} \cdot D_{18m}^{OIS} + F_{18m,2y} \cdot \alpha_{18m,2y} \cdot D_{2y}^{OIS}\bigr) = 0
$$

Como todos los factores de descuento OIS se conocen de la curva Eonia, **solo las tasas forward son desconocidas.** 

El extremo corto de la curva Euribor se deriva de instrumentos al contado, de modo que las dos primeras tasas forward **están dadas**. 

Luego, interpolando entre la tasa forward a un año y la de 18 meses para obtener $F_{1y,18m}$, como función de $F_{6m,1y}$ y $F_{18m,2y}$, solo queda la última tasa forward por determinar. Así podemos, nuevamente, aplicar un procedimiento paso a paso para obtener todas las tasas forward y finalmente podemos valuar un swap Euribor colateralizado.

La curva forward implícita obtenida de este modo difiere ligeramente de las tasas forward calculadas la curva zero directamente ... *ojito ahí*. Esto se debe a que, aunque ambas patas se descuenten con la misma curva *ahora más baja*, la distribución de los flujos en el tiempo no es exactamente igual. 

Como resultado, la curva forward obtenida se recalibra usando tasas implícitas de las cotizaciones de swaps. Mercurio (2009) tiene en cuenta esta diferencia mediante el uso de contratos de forward rate agreements para determinar las tasas forward y luego descontarlas con la curva OIS.
